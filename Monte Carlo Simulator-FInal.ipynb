```python
#GET 10 YEARS OF CLOSING PRICES]
#Load past daily closing prices into Pandas
```

```python
import yfinance as yf
NSEI = yf.download("^NSEI",start = "2006-01-01", end = "2026-01-01")
NSEI
NSEI.columns = ["Close", "High", "Low", "Open", "Volume"]
NSEI
```

```python
#CONVERT TO DAILY PERCENTAGE CHANGES]
#│Use .pct_change().dropna() so you have a clean list
#│of ~2,500 daily percentage moves (your "hat of slips")
```

```python
import numpy as np
import pandas as pd
close = pd.DataFrame(NSEI["Close"])
close["PCT_Change"] = close.pct_change()
close = close.dropna()
close
```

```python
#SPIN 10,000 FUTURES × 252 DAYS AT ONCE]
#│Use np.random.choice(past_returns, size=(10000, 252), replace=True)
#│No loops needed—creates a 10,000 × 252 grid of daily moves
```

```python
nextdays = np.random.choice(close["PCT_Change"], size = (252,10000), replace = True)
nextdays = pd.DataFrame(nextdays)
nextdays
```

```python
#Before starting remove NaN value from the random generator as it could be put up in any prediction making returns 0
#Remove *100 and keep decimals cuz in compounding it would create a percentage error
```

```python
#COMPOUND ACROSS THE 252 DAYS]
#│Use np.cumprod(1 + random_returns, axis=1)
#│Starts at 1.0 and multiplies day-by-day across each row
```

```python
cmpd = np.cumprod(1+nextdays, axis = 0)
cmpd
```

```python
#CONVERT TO FINAL PERCENTAGE RETURNS]
#│Take the last column (Day 252) and subtract 1:
#│final_returns = growth_paths[:, -1] - 1
```

```python
yrly_returns = pd.DataFrame(cmpd.iloc[-1]-1)
yrly_returns
```

```python
#MEASURE THE ODDS & RISK]
#│From those 10,000 final 1-year return percentages, check:
#│• Median return (50th percentile)
#│• Worst 5% return (5th percentile) & Worst 1% return (1st percentile)
#│• % of simulated years that end negative (< 0%)
```

```python
#Skewness: Measure of which side the data is concentrated more comparing it to median
#Positive: Mean > Median, means crazy bull years pulled the number up; More Bullish: So investing can be considered a good choice
#Negative: Median > Mean, means crazy bear years were there; So investing should be considered and talked about thoughtfully.
#Median: Tells about the average year looks like, cuz its at the middle of those 10,000 values, above it all bullish and below it all bearish
#Compare the top 5th percentile(95 percentile) and bottom 5 percentile to understand about whether the risk is worth taking or not
#Probability of loss, basically tells us what are the chances of losing the money (Out of those 10,000: how many ended below 0%)
#We dont use mean to find about average year returns as extreme outliers tend to shift it
#VaR is calculated post sorting and looking at bottom values to understand what is the worst that could happen
```

```python
yrly_returns = yrly_returns.sort_values(by=yrly_returns.columns[0], ascending=False)
```

```python
yrly_returns
```

```python
var_95 = yrly_returns.iloc[-500]*100
var_99 = yrly_returns.iloc[-100]*100
pct_95 = yrly_returns.iloc[500]*100
median_return = np.median(yrly_returns)*100
mean_return = np.mean(yrly_returns)*100

if mean_return > median_return:
    print("Positive Skewed")
else:
    print("Negatively Skewed")

loss = int((yrly_returns < 0).sum().sum())

print("Lowest 5th percentile:", round(float(var_95.iloc[0]), 5))
print("Lowest 1st percentile:", round(float(var_99.iloc[0]), 6))
print("Highest 5th percentile:", round(float(pct_95.iloc[0]), 6))
print("Median return:", round(float(median_return), 6))
print("Mean return:", round(float(mean_return), 6))
print("Probability of a loss is",loss/10000)
if mean_return > median_return:
    print("Positive Skewed")
else:
    print("Negatively Skewed")
```
