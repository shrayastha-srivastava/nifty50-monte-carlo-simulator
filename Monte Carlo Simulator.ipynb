```python
#GET 10 YEARS OF CLOSING PRICES]
#Load past daily closing prices into Pandas
```

```python
import yfinance as yf
NSEI = yf.download("^NSEI",start = "2006-01-01", end = "2026-01-01")
NSEI
NSEI.columns = ["Close", "High", "Low", "Open", "Volume"]
NSEI
```

```python
#CONVERT TO DAILY PERCENTAGE CHANGES]
#│Use .pct_change().dropna() so you have a clean list
#│of ~2,500 daily percentage moves (your "hat of slips")
```

```python
import numpy as np
import pandas as pd
close = pd.DataFrame(NSEI["Close"])
close["PCT_Change"] = close.pct_change()
close = close.dropna()
close
```

```python
#SPIN 10,000 FUTURES × 252 DAYS AT ONCE]
#│Use np.random.choice(past_returns, size=(10000, 252), replace=True)
#│No loops needed—creates a 10,000 × 252 grid of daily moves
```

```python
nextdays = np.random.choice(close["PCT_Change"], size = (252,10000), replace = True)
nextdays = pd.DataFrame(nextdays)
nextdays
```

```python
#Before starting remove NaN value from the random generator as it could be put up in any prediction making returns 0
#Remove *100 and keep decimals cuz in compounding it would create a percentage error
```

```python
#COMPOUND ACROSS THE 252 DAYS]
#│Use np.cumprod(1 + random_returns, axis=1)
#│Starts at 1.0 and multiplies day-by-day across each row
```

```python
cmpd = np.cumprod(1+nextdays, axis = 0)
cmpd
```

```python
#CONVERT TO FINAL PERCENTAGE RETURNS]
#│Take the last column (Day 252) and subtract 1:
#│final_returns = growth_paths[:, -1] - 1
```

```python
yrly_returns = pd.DataFrame(cmpd.iloc[-1]-1)
yrly_returns
```

```python
#MEASURE THE ODDS & RISK]
#│From those 10,000 final 1-year return percentages, check:
#│• Median return (50th percentile)
#│• Worst 5% return (5th percentile) & Worst 1% return (1st percentile)
#│• % of simulated years that end negative (< 0%)
```

```python
#Skewness: Measure of which side the data is concentrated more comparing it to median
#Positive: Mean > Median, means crazy bull years pulled the number up; More Bullish: So investing can be considered a good choice
#Negative: Median > Mean, means crazy bear years were there; So investing should be considered and talked about thoughtfully.
#Median: Tells about the average year looks like, cuz its at the middle of those 10,000 values, above it all bullish and below it all bearish
#Compare the top 5th percentile(95 percentile) and bottom 5 percentile to understand about whether the risk is worth taking or not
#Probability of loss, basically tells us what are the chances of losing the money (Out of those 10,000: how many ended below 0%)
#We dont use mean to find about average year returns as extreme outliers tend to shift it
#VaR is calculated post sorting and looking at bottom values to understand what is the worst that could happen
```

```python
yrly_returns = yrly_returns.sort_values(by=yrly_returns.columns[0], ascending=False)
```

```python
np.random.seed(13)
predicted_var95 = []
predicted_var99 = []
breach_var95 = []
breach_var99 = []
dates = []
pct_returns = np.array(pct_returns).flatten()
for i in range(3730):
    hat = pct_returns[0+i:756+i]
    predict = np.random.choice(hat, size = 10000, replace = True)
    predict = np.sort(predict)
    predicted_var95.append(predict[499])
    predicted_var99.append(predict[99])
    if pct_returns[756+i]>predict[499]:
        breach_var95.append(False)
    else:
        breach_var95.append(True)
    if pct_returns[756+i] > predict[99]:
        breach_var99.append(False)
    else:
        breach_var99.append(True)


data = pd.DataFrame()
data["Predicted_Var95"] = predicted_var95
data["Predicted_Var99"] = predicted_var99
data["Breach_Var95"] = breach_var95
data["Breach_Var99"] = breach_var99
data

trueval_var95 = np.sum(data["Breach_Var95"] == True)
trueval_var99 = np.sum(data["Breach_Var99"] == True)
print("Breach Rate of VaR95 is", (trueval_var95)*100/3730,"Breach Rate of VaR99 is", (trueval_var99)*100/3730 )
```

```python
#Christoffersen's Independence Test
```

```python
#Breach_Var95
n00 = 0
n01 = 0
n10 = 0
n11 = 0
for i in range(1,len(breach_var95)):
        if breach_var95[i-1] == False and breach_var95[i] == False:
            n00+=1
        elif breach_var95[i-1] == False and breach_var95[i] == True:
            n01+=1
        elif breach_var95[i-1] == True and breach_var95[i] == False:
            n10+=1
        elif breach_var95[i-1] == True and breach_var95[i] == True:
            n11+=1
pi0 = n01/(n00+n01)   #Used to calculate if today was a breach given yesterday was not
pi1 = n11/(n11+n10)   #Used to calculate if today was a breach given yesterday was also a breach
pi = (n01+n11)/(n00+n01+n10+n11) #Used to calculate overall breach probability



import math
log_L0 = (n00 + n10) * math.log(1 - pi) + (n01 + n11) * math.log(pi)
log_L1 = n00 * math.log(1 - pi0) + n01 * math.log(pi0) + n10 * math.log(1 - pi1) + n11 * math.log(pi1)
LR_ind = -2 * (log_L0 - log_L1) # Christoffersen Independence Statistic (chi-square, df=1, 95% critical value = 3.841)
# - LR_ind < 2.706: Breaches are independent (pi_0 ≈ pi_1); model adapts well.
# - 2.706 to 3.841: Borderline / mild spillover, passes strict 95% cutoff.
# - LR_ind > 3.841: FAILS independence at 95% confidence (pi_1 >> pi_0); model reacts too slowly and suffers breach clustering.
LR_ind
```

```python
#Breach_Var99
n_00 = 0
n_01 = 0
n_10 = 0
n_11 = 0
for i in range(1,len(breach_var99)):
        if breach_var99[i-1] == False and breach_var99[i] == False:
            n_00+=1
        elif breach_var99[i-1] == False and breach_var99[i] == True:
            n_01+=1
        elif breach_var99[i-1] == True and breach_var99[i] == False:
            n_10+=1
        elif breach_var99[i-1] == True and breach_var99[i] == True:
            n_11+=1
pi_0 = n_01/(n_00+n_01)   #Used to calculate if today was a breach given yesterday was not
pi_1 = n_11/(n_11+n_10)   #Used to calculate if today was a breach given yesterday was also a breach
pi_ = (n_01+n_11)/(n_00+n_01+n_10+n_11) #Used to calculate overall breach probability



import math
log_L_0 = (n_00 + n_10) * math.log(1 - pi_) + (n_01 + n_11) * math.log(pi_)
log_L_1 = n_00 * math.log(1 - pi_0) + n_01 * math.log(pi_0) + n_10 * math.log(1 - pi_1) + n_11 * math.log(pi_1)
LR_ind_ = -2 * (log_L_0 - log_L_1) # Christoffersen Independence Statistic (chi-square, df=1, 95% critical value = 3.841)
# - LR_ind < 2.706: Breaches are independent (pi_0 ≈ pi_1); model adapts well.
# - 2.706 to 3.841: Borderline / mild spillover, passes strict 95% cutoff.
# - LR_ind > 3.841: FAILS independence at 95% confidence (pi_1 >> pi_0); model reacts too slowly and suffers breach clustering.
LR_ind_
```
